```python
from astropy.io import fits, ascii
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import sparc4.product_plots as s4plt
import sparc4.pipeline_lib as s4pipelib
from mpl_toolkits.mplot3d import Axes3D
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

# Canal 1

```python
pasta='/media/fernando/KINGSTON/Doutorado/2025/reduced'
night='20250325'
object='V1025Cen'

apertures=[10]

channel='1'

polar_stack_example = f'{pasta}/{night}/sparc4acs{channel}/{night}_s4c{channel}_Gain2_1.0_Conventional_2_{object}_POLAR_L4_stack.fits'
polar_example =       f'{pasta}/{night}/sparc4acs{channel}/{night}_s4c{channel}_Gain2_1.0_Conventional_2_{object}_POLAR_L4_polar.fits'
```

```python
s4plt.plot_sci_polar_frame(polar_stack_example)
```

```python
# Abre o arquivo FITS
with fits.open(polar_example) as hdul:
    i = 2  # Índice do objeto
    num_fo = 16  # FO0000 a FO0015
    aberturas =  range(1, len(apertures))  # hdul[1] até hdul[20], correspondem às aberturas 2 a 21
    
    # Eixo x agora representa as aberturas reais
    x = np.array([j + 1 for j in aberturas])  # 2 a 21

    plt.figure(figsize=(14, 7))

    # Para cada variável FOxxxx e FExxxx (0 a 15), coletamos os valores ao longo das aberturas
    for j in range(num_fo):
        fo_vals = [hdul[h].data[f'FO{j:04d}'][i] for h in aberturas]
        efo_vals = [hdul[h].data[f'EFO{j:04d}'][i] for h in aberturas]
        fe_vals = [hdul[h].data[f'FE{j:04d}'][i] for h in aberturas]
        efe_vals = [hdul[h].data[f'EFE{j:04d}'][i] for h in aberturas]

        plt.errorbar(x, fo_vals, yerr=efo_vals, fmt='-', capsize=2, alpha=0.7,
                     label=f'FO{j:04d}')
        plt.errorbar(x, fe_vals, yerr=efe_vals, fmt='--', capsize=2, alpha=0.7,
                     label=f'FE{j:04d}')

    plt.xlabel('Abertura (número real, de 2 a 21)')
    plt.ylabel('Valor')
    plt.title(f'Curvas de FO e FE para srcindex = {i} (posição {i+1})')
    plt.grid(True)
    plt.legend(ncol=2, fontsize='small', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
```
