## Fotometria

```python
from astropy.io import fits
import pandas as pd
import os
import numpy as np
import sparc4.product_plots as s4plt


# === CONFIGURAÇÕES GERAIS ===
path = '/media/fernando/KINGSTON/Doutorado/2025/reduced/'
date = '20250707'
object = 'V462Lup'

# === LOOP SOBRE OS QUATRO CANAIS ===
for channel in range(1, 5):

    print(f"\n\n{'='*70}")
    print(f"Processando canal {channel} (fotometria)")
    print(f"{'='*70}")

    phot_fits_file = f'{path}{date}/sparc4acs{channel}/{date}_s4c{channel}_{object}_POLAR_L4_S+N_lc.fits'
    polar_stack_example = f'{path}{date}/sparc4acs{channel}/{date}_s4c{channel}_Gain2_1.0_Conventional_2_{object}_POLAR_L4_stack.fits'

    if not os.path.exists(phot_fits_file):
        print(f"Arquivo não encontrado: {phot_fits_file}")
        continue

    # Abre o arquivo FITS
    hdul = fits.open(phot_fits_file)

    # Lista extensões disponíveis
    available_exts = [hdu.name for hdu in hdul if isinstance(hdu.data, fits.FITS_rec)]
    print("\nExtensões disponíveis (aberturas):")
    for ext in available_exts:
        print(" -", ext)

    # === Escolha da abertura ===
    aperture = input("\nDigite a abertura desejada (por ex.: 005, 008, 010, etc.): ").strip()
    ext_name = f"CATALOG_PHOT_AP{aperture.zfill(3)}"

    if ext_name not in available_exts:
        print(f"Abertura {ext_name} não encontrada neste arquivo.")
        hdul.close()
        continue

    # === Lê a tabela da extensão escolhida ===
    data = hdul[ext_name].data

#=======================================
    
    # Exemplo para verificar antes e depois
    print("Exemplo antes de byteswap:")
    print(data[0])  # Printando o primeiro item
    
    # ⚠️ Conversão para little-endian
#    data = data.byteswap().newbyteorder()
#    data = data.byteswap().astype(data.dtype.newbyteorder('<'))

    # Converte o recarray para um ndarray simples
    data = data.view(np.ndarray)

    # Inverte a ordem dos bytes e ajusta o byte-order para little-endian
    data = data.byteswap().view(data.dtype.newbyteorder('<'))
    
    print("Exemplo depois de byteswap:")
    print(data[0])  # O valor deve ser idêntico
    
#=======================================

    # Cria o DataFrame com dados já convertidos
    df = pd.DataFrame(np.array(data))

    s4plt.plot_sci_polar_frame(polar_stack_example)

    # Lista os SRCINDEX disponíveis
    unique_src = df['SRCINDEX'].unique()
    print("\nSRCINDEX disponíveis:", unique_src)

    # === Escolha do SRCINDEX principal ===
    srcindex_main = int(input("\nDigite o SRCINDEX principal desejado: "))

    if srcindex_main not in unique_src:
        print(f"SRCINDEX {srcindex_main} não encontrado nesta abertura.")
        hdul.close()
        continue

    # === Filtra dados do SRCINDEX principal ===
    df_main = df[df['SRCINDEX'] == srcindex_main][['TIME', 'SRCINDEX', 'MAG', 'EMAG']]
    df_main = df_main.rename(columns={
        'SRCINDEX': f'SRCINDEX_{srcindex_main}',
        'MAG': f'MAG_{srcindex_main}',
        'EMAG': f'EMAG_{srcindex_main}'
    })

    # === ALTERAÇÃO 1 ===
    # Pergunta se o usuário quer adicionar mais SRCINDEX
    additional_srcs = []
    for i in range(3):  # até 3 adicionais (9 colunas)
        ans = input(f"\nDigite outro SRCINDEX (ou pressione Enter para parar): ").strip()
        if ans == "":
            break
        try:
            src = int(ans)
            if src not in unique_src:
                print(f"SRCINDEX {src} não encontrado. Ignorando.")
                continue
            additional_srcs.append(src)
        except ValueError:
            print("Entrada inválida. Ignorando.")
            continue

    # === ALTERAÇÃO 2 ===
    # Começa com o DataFrame principal
    df_merged = df_main.copy()

    # Faz merge (inner join por TIME) com cada SRCINDEX adicional
    for src in additional_srcs:
        df_aux = df[df['SRCINDEX'] == src][['TIME', 'SRCINDEX', 'MAG', 'EMAG']]
        df_aux = df_aux.rename(columns={
            'SRCINDEX': f'SRCINDEX_{src}',
            'MAG': f'MAG_{src}',
            'EMAG': f'EMAG_{src}'
        })
        df_merged = pd.merge(df_merged, df_aux, on='TIME', how='left')

    # === Cria diretório de saída ===
    output_dir = os.path.join(path, date, 'export_txt')
    os.makedirs(output_dir, exist_ok=True)

    # === Nome do arquivo de saída ===
    all_srcs = [srcindex_main]# + additional_srcs
    src_label = "_".join([f"SRC{src}" for src in all_srcs])
    output_file = os.path.join(output_dir, f'{object}_ch{channel}_AP{aperture}_{src_label}.txt')

    # === Salva como arquivo texto ===
    df_merged.to_csv(output_file, sep='\t', index=False, float_format='%.10f', na_rep='     NaN     ')


    print(f"\nArquivo salvo: {output_file}")

    # Fecha o FITS
    hdul.close()

print("\nProcessamento concluído para todos os canais.")
```

# Polarimetria

```python
from astropy.io import fits
import pandas as pd
import os
import numpy as np
import sparc4.product_plots as s4plt

# === CONFIGURAÇÕES GERAIS ===
path = '/media/fernando/KINGSTON/Doutorado/2025/reduced/'
date = '20250707'
object = 'V462Lup'

# === LOOP SOBRE OS QUATRO CANAIS ===
for channel in range(1, 5):

    print(f"\n\n{'='*70}")
    print(f"Processando canal {channel} (polarimetria)")
    print(f"{'='*70}")

    pol_fits_file = f'{path}{date}/sparc4acs{channel}/{date}_s4c{channel}_{object}_POLAR_L4_ts.fits'
    polar_stack_example = f'{path}{date}/sparc4acs{channel}/{date}_s4c{channel}_Gain2_1.0_Conventional_2_{object}_POLAR_L4_stack.fits'

    if not os.path.exists(pol_fits_file):
        print(f"Arquivo não encontrado: {pol_fits_file}")
        continue

    # Abre o arquivo FITS
    hdul = fits.open(pol_fits_file)

    # Garante que a extensão 'POLAR_TIMESERIES' existe
    if 'POLAR_TIMESERIES' not in [hdu.name for hdu in hdul]:
        print("Extensão 'POLAR_TIMESERIES' não encontrada neste arquivo.")
        hdul.close()
        continue

    # Lê a tabela
    data = hdul['POLAR_TIMESERIES'].data

#=======================================
    
    # Exemplo para verificar antes e depois
    print("Exemplo antes de byteswap:")
    print(data[0])  # Printando o primeiro item
    
    # ⚠️ Conversão para little-endian
#    data = data.byteswap().newbyteorder()
#    data = data.byteswap().astype(data.dtype.newbyteorder('<'))

    # Converte o recarray para um ndarray simples
    data = data.view(np.ndarray)

    # Inverte a ordem dos bytes e ajusta o byte-order para little-endian
    data = data.byteswap().view(data.dtype.newbyteorder('<'))
    
    print("Exemplo depois de byteswap:")
    print(data[0])  # O valor deve ser idêntico
    
#=======================================

    # Cria o DataFrame
    df = pd.DataFrame(np.array(data))

    s4plt.plot_sci_polar_frame(polar_stack_example)

    # Lista SRCINDEX disponíveis
    unique_src = df['SRCINDEX'].unique()
    print("\nSRCINDEX disponíveis:", unique_src)

    # Escolha do SRCINDEX
    srcindex = int(input("\nDigite o SRCINDEX desejado: "))

    if srcindex not in unique_src:
        print(f"SRCINDEX {srcindex} não encontrado.")
        hdul.close()
        continue

    # Seleciona colunas de interesse
    selected_cols = [
        'TIME', 'SRCINDEX', 'MAG', 'EMAG',
        'Q', 'EQ', 'U', 'EU', 'V', 'EV',
        'P', 'EP', 'THETA', 'ETHETA'
    ]

    # Verifica se todas as colunas estão presentes
    missing = [col for col in selected_cols if col not in df.columns]
    if missing:
        print("Aviso: as seguintes colunas não foram encontradas:", missing)

    # Filtra e seleciona colunas
    df_filtered = df[df['SRCINDEX'] == srcindex][selected_cols]

    # Diretório de saída
    output_dir = os.path.join(path, date, 'export_txt')
    os.makedirs(output_dir, exist_ok=True)

    # Nome do arquivo de saída
    output_file = os.path.join(output_dir, f'{object}_ch{channel}_SRC{srcindex}_polar.txt')

    # Salva como arquivo texto
    df_filtered.to_csv(output_file, sep='\t', index=False, float_format='%.10f', na_rep='     NaN     ')

    print(f"\nArquivo salvo: {output_file}")

    # Fecha o FITS
    hdul.close()

print("\nProcessamento concluído para todos os canais (polarimetria).")
```
