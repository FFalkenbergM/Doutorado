```python
import os
import pandas as pd

# === CONFIGURAÇÕES GERAIS ===
# Caminho base onde ficam as pastas de data:
base_path = "/media/fernando/KINGSTON/Doutorado/2025/reduced"

# >>>>>> COLOQUE AQUI O NOME DA PASTA DE DATA <<<<<<
date_folder = "20250706"

# Valor a ser somado em cada valor da coluna MAG:
correcao_mag = 3.0102999570

# Caminho completo para a pasta export_txt:
export_path = os.path.join(base_path, date_folder, "export_txt")

# === VERIFICAÇÕES INICIAIS ===
if not os.path.exists(export_path):
    print(f"ERRO: pasta não encontrada -> {export_path}")
else:
    print(f"Pasta selecionada: {export_path}")

    # Busca todos os arquivos que terminam com *polar.txt
    txt_files = [f for f in os.listdir(export_path) if f.endswith("polar.txt")]

    if not txt_files:
        print("Nenhum arquivo que termine com '*polar.txt' foi encontrado.")
    else:
        print(f"{len(txt_files)} arquivo(s) encontrado(s):")
        for f in txt_files:
            print(" -", f)

        # === PROCESSAMENTO DOS ARQUIVOS ===
        for fname in txt_files:
            file_path = os.path.join(export_path, fname)
            try:
                # Lê o arquivo
                df = pd.read_csv(file_path, delim_whitespace=True, comment='#')

                # Verifica se já foi corrigido
                if "MAG_corr" in df.columns:
                    print(f"[PULADO] {fname} já contém coluna MAG_corr.")
                    continue

                if "MAG" not in df.columns:
                    print(f"[ERRO] {fname} não possui coluna MAG.")
                    continue

                # Cria uma nova coluna MAG_corr com os valores corrigidos
                df["MAG_corr"] = df["MAG"] + correcao_mag

                # Recria a ordem original, substituindo MAG por MAG_corr
                colunas_ordenadas = []
                for c in df.columns:
                    if c == "MAG":
                        colunas_ordenadas.append("MAG_corr")
                    elif c != "MAG_corr":
                        colunas_ordenadas.append(c)
                df = df[colunas_ordenadas]

                # Remove a antiga coluna MAG
                if "MAG" in df.columns:
                    df.drop(columns=["MAG"], inplace=True)

                # Salva o arquivo sobrescrevendo o original
                df.to_csv(file_path, sep="\t", index=False, float_format="%.10f")
                print(f"[OK] {fname} corrigido e salvo.")
            except Exception as e:
                print(f"[ERRO] ao processar {fname}: {e}")
```
